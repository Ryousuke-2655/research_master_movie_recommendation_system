```python
import numpy as np
import pandas as pd

pd.set_option('display.max_columns',None)
pd.set_option('display.max_rows',None)
pd.set_option('display.max_colwidth', None)
base_path = "/content/drive/MyDrive/研究用コード/一時保存"
```

```python
df_details = pd.read_csv(f'{base_path}/tmdb_movies_details_1960_2025.csv',encoding ='utf-8' )
df_details.info()
```

```python
df_overview = df_details[["id","overview"]]
df_overview["overview_count"] = df_overview["overview"].str.len()
df_overview.describe()
```

```python
import matplotlib.pyplot as plt
import japanize_matplotlib

fig = plt.figure()
ax = fig.add_subplot(1,1,1)

ax.hist(df_overview["overview_count"], bins=50)
ax.set_title("映画概要文の長さ")
ax.set_xlabel('overview_len')
ax.set_ylabel('count')
fig.show()
```
